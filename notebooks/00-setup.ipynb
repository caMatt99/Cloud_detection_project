````python
'''
import os
import shutil
import random

# Percorsi
BASE        = '/content/drive/MyDrive/cloud_detection/dataset'
TRAIN_PATH  = f'{BASE}/train'
VAL_PATH    = f'{BASE}/val'

# Proporzioni del paper: train 49%, val 21%, test 30%
# Il validation è il 21/70 = 30% del training set attuale
VAL_RATIO = 0.30

random.seed(42)  # per riproducibilità

for classe in ['true', 'false']:
    src_dir = f'{TRAIN_PATH}/{classe}'
    val_dir = f'{VAL_PATH}/{classe}'
    os.makedirs(val_dir, exist_ok=True)

    files = os.listdir(src_dir)
    random.shuffle(files)

    n_val = int(len(files) * VAL_RATIO)
    val_files = files[:n_val]

    for f in val_files:
        shutil.move(f'{src_dir}/{f}', f'{val_dir}/{f}')

    print(f"Classe '{classe}': {len(files)} totali → {n_val} spostati in val, {len(files)-n_val} rimasti in train")

print("\nDone. Struttura finale:")
for split in ['train', 'val', 'test']:
    for classe in ['true', 'false']:
        n = len(os.listdir(f'{BASE}/{split}/{classe}'))
        print(f"  {split}/{classe}: {n} immagini")

'''
````
